# Trip Planner

This beginner Python program calculates the total cost of a group trip.

It uses the destination, number of people, number of nights, and accommodation type. It also applies the group and long-stay discounts.

```python

transport_prices = {
    "pokhara": 3000,
    "chitwan": 2500,
    "kathmandu": 1500
}


accommodation_prices = {
    "budget": 1000,
    "standard": 2000,
    "luxury": 4000
}

print("Trip prices are ready.")
```

```python
destination = input("Destination (Pokhara/Chitwan/Kathmandu): ").strip().lower()
people = int(input("Number of people: "))
nights = int(input("Number of nights: "))
accommodation = input("Accommodation (Budget/Standard/Luxury): ").strip().lower()
```

```python

if destination not in transport_prices:
    print("Invalid destination. Please choose Pokhara, Chitwan, or Kathmandu.")
elif accommodation not in accommodation_prices:
    print("Invalid accommodation. Please choose Budget, Standard, or Luxury.")
elif people <= 0 or nights <= 0:
    print("Number of people and nights must be greater than zero.")
else:
  
    transport_cost = transport_prices[destination] * people
    accommodation_cost = accommodation_prices[accommodation] * people * nights
    total_before_discount = transport_cost + accommodation_cost

    discount_rate = 0

    if people >= 4:
        discount_rate += 0.10

    if nights >= 5:
        discount_rate += 0.05


    discount_rate = min(discount_rate, 0.15)

    discount_amount = total_before_discount * discount_rate
    final_trip_cost = total_before_discount - discount_amount
    cost_per_person = final_trip_cost / people


    print("\n--- TRIP COST SUMMARY ---")
    print(f"Destination: {destination.title()}")
    print(f"Number of people: {people}")
    print(f"Number of nights: {nights}")
    print(f"Accommodation: {accommodation.title()}")
    print(f"Transport cost: Rs. {transport_cost:,.2f}")
    print(f"Accommodation cost: Rs. {accommodation_cost:,.2f}")
    print(f"Total before discount: Rs. {total_before_discount:,.2f}")
    print(f"Discount: {discount_rate * 100:.0f}%")
    print(f"Discount amount: Rs. {discount_amount:,.2f}")
    print(f"Final trip cost: Rs. {final_trip_cost:,.2f}")
    print(f"Cost per person: Rs. {cost_per_person:,.2f}")
```
